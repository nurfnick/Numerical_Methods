```python
import sympy

x = sympy.Symbol('x')

f = sympy.cos(x) - x

```

```python
f.diff(x)
```

```python
def newton(x0,f):
  temp = x0
  for i in range(5):
  #while f.subs(x,temp)>epsilon:
    temp = temp-f.subs(x,temp)/f.diff(x).subs(x,temp)
    print(sympy.N(temp),sympy.N(f.subs(x,temp)))
  return sympy.N(temp)
```
