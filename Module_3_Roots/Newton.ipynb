```python
import sympy

x = sympy.Symbol('x')

f = x**2-2

```

```python
f.diff(x).subs(x,2)
```

```python
def newton(x0,f,epsilon):
  temp = x0
  for i in range(10):
  #while f.subs(x,temp)>epsilon:
    temp = temp-f.subs(x,temp)/f.diff(x).subs(x,temp)
    print(sympy.N(temp))
  return temp
```
