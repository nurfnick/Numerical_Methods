```python
def fixedPoint(a,f):
  pt = a
  for i in range(20):
    pt = f(pt)
    print(pt)
  return pt
```

Solving $0= x^3-x-1$

```python
def f(x):
  return (x+1)**(1/3)
```

```python
y = fixedPoint(1,f)
y
```

```python
f(y)-y
```
