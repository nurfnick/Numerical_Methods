```python
def bisection(a,b,f):
  left = a
  right = b
  for i in range(0,10):
    if f((left+right)/2) == 0:
      return (left+right)/2
    elif f(left)*f(((left+right)/2))<0:
      right = (left+right)/2
    elif f(right)*f((left+right)/2)<0:
      left = (left+right)/2
    else:
      return "Somethings broken"
  return (left,right)
```

```python
def f(x):
  return x**2-2
```

```python
bisection(1,2,f)
```
