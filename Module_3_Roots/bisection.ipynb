```python
def bisection(a,b,f):
  left = a
  right = b
  for i in range(0,10):
    if f((left+right)/2) == 0:
      return (left+right)/2
    elif f(left)*f(((left+right)/2))<0:
      right = (left+right)/2
    elif f(right)*f((left+right)/2)<0:
      left = (left+right)/2
    else:
      return "Somethings broken"
  return (left,right)
```

```python
def f(x):
  return x**2-2
```

```python
bisection(1,2,f)
```

```python
def f(x):
  return x**3-x-1
```

```python
def bisectionwithTol(a,b,f,tol):
  left = a
  right = b
  n=1
  while (right-left)/2 >tol:
    if f((left+right)/2) == 0:
      return (left+right)/2
    elif f(left)*f(((left+right)/2))<0:
      right = (left+right)/2
    elif f(right)*f((left+right)/2)<0:
      left = (left+right)/2
    else:
      return "Somethings broken"
    print(n,left,right)
    n+=1
  return (left+right)/2
```

```python
bisectionwithTol(0,2,f,0.0000001)
```
