# Numerical Approximation of the Error Function (erf)

This notebook demonstrates how to approximate the error function using Riemann sums (Left, Right, and Midpoint/Center endpoints) to estimate the area under the curve of $f(t) = e^{-t^2}$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erf
import ipywidgets as widgets
from ipywidgets import interact

def plot_erf_approximation(x_max, N, method):
    # Define the integrand for erf: f(t) = (2 / sqrt(pi)) * e^(-t^2)
    def f(t):
        return (2.0 / np.sqrt(np.pi)) * np.exp(-t**2)

    # Set up the partition
    t_endpoints = np.linspace(0, x_max, N + 1)
    dt = x_max / N

    # Determine evaluation points based on the chosen method
    if method == 'Left':
        t_eval = t_endpoints[:-1]
    elif method == 'Right':
        t_eval = t_endpoints[1:]
    elif method == 'Center':
        t_eval = (t_endpoints[:-1] + t_endpoints[1:]) / 2.0

    # Calculate approximation (Riemann sum)
    heights = f(t_eval)
    approximation = np.sum(heights * dt)
    actual_val = erf(x_max)
    error = np.abs(approximation - actual_val)

    # Plot setup
    plt.figure(figsize=(10, 6))

    # Draw smooth curve representing the integrand
    t_smooth = np.linspace(0, max(x_max, 2.5), 300)
    plt.plot(t_smooth, f(t_smooth), 'r-', lw=2, label=r'$f(t) = \frac{2}{\sqrt{\pi}} e^{-t^2}$')

    # Plot the approximating rectangles
    for i in range(N):
        t_left = t_endpoints[i]
        t_right = t_endpoints[i+1]
        h = heights[i]

        # Draw rectangle
        plt.bar(t_left, h, width=dt, align='edge', alpha=0.3, edgecolor='blue', color='skyblue')

        # Mark evaluation point
        plt.plot(t_eval[i], h, 'bo', markersize=4)

    plt.title(f'Approximating erf({x_max:.2f}) using {method} Endpoints (N={N})')
    plt.xlabel('t')
    plt.ylabel('f(t)')
    plt.xlim(0, max(x_max * 1.2, 1.5))
    plt.ylim(0, 1.2)
    plt.grid(True, linestyle='--', alpha=0.5)

    # Add metrics text box
    textstr = '\n'.join((
        f'Approximation: {approximation:.6f}',
        f'Actual erf({x_max:.2f}): {actual_val:.6f}',
        f'Absolute Error: {error:.6f}'
    ))
    props = dict(boxstyle='round', facecolor='wheat', alpha=0.5)
    plt.gca().text(0.05, 0.95, textstr, transform=plt.gca().transAxes, fontsize=11,
                    verticalalignment='top', bbox=props)

    plt.legend(loc='upper right')
    plt.show()

# Create interactive controls
interact(
    plot_erf_approximation,
    x_max=widgets.FloatSlider(value=1.5, min=0.1, max=3.0, step=0.1, description='Upper Limit (x)'),
    N=widgets.IntSlider(value=10, min=1, max=100, step=1, description='Intervals (N)'),
    method=widgets.Dropdown(options=['Left', 'Right', 'Center'], value='Center', description='Method')
);
```
